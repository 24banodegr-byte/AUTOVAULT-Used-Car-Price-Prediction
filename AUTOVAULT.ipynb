# MODEL COMPARISON AND ACCURACY EVALUATION

```python
# ============================================================
# MODEL COMPARISON AND ACCURACY EVALUATION
# ============================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import (
    RandomForestRegressor,
    ExtraTreesRegressor,
    GradientBoostingRegressor
)
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Separate input features (X) and target variable (y)
X = df.drop(columns=["selling_price"])
y = df["selling_price"]

# Identify categorical and numerical columns
categorical_cols = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

numerical_cols = X.select_dtypes(
    include=np.number
).columns.tolist()

# Handle missing values
X = X.copy()

for col in categorical_cols:
    X[col] = X[col].fillna("Unknown").astype(str)

for col in numerical_cols:
    X[col] = X[col].fillna(X[col].median())

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("categorical", OneHotEncoder(handle_unknown="ignore"),
         categorical_cols),
        ("numerical", "passthrough", numerical_cols)
    ]
)

# Models for comparison
models = {
    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),
    "Extra Trees": ExtraTreesRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),
    "Gradient Boosting": GradientBoostingRegressor(
        random_state=42
    )
}

# Train, predict and evaluate each model
results = []

for model_name, model in models.items():

    pipeline = Pipeline([
        ("preprocessing", preprocessor),
        ("model", model)
    ])

    pipeline.fit(X_train, y_train)
    predictions = pipeline.predict(X_test)

    mae = mean_absolute_error(y_test, predictions)
    rmse = np.sqrt(mean_squared_error(y_test, predictions))
    r2 = r2_score(y_test, predictions)

    # Percentage of predictions within +/-10% of actual price
    actual = np.asarray(y_test)
    predicted = np.asarray(predictions)

    valid = actual != 0

    within_10_percent = (
        np.mean(
            np.abs(predicted[valid] - actual[valid])
            / np.abs(actual[valid]) <= 0.10
        ) * 100
        if valid.any()
        else np.nan
    )

    results.append({
        "Model": model_name,
        "MAE": round(mae, 2),
        "RMSE": round(rmse, 2),
        "R2 Score": round(r2, 4),
        "Within +/-10% (%)": round(within_10_percent, 2)
    })

# Display comparison table
comparison_df = pd.DataFrame(results)
comparison_df = comparison_df.sort_values(
    by="RMSE",
    ascending=True
).reset_index(drop=True)

print("\nMODEL COMPARISON AND ACCURACY EVALUATION")
print("=" * 65)
print(comparison_df.to_string(index=False))

# Identify the best model based on RMSE
best_model = comparison_df.iloc[0]

print("\nBEST PERFORMING MODEL")
print("=" * 35)
print("Model:", best_model["Model"])
print("RMSE:", best_model["RMSE"])
print("MAE:", best_model["MAE"])
print("R2 Score:", best_model["R2 Score"])
```

The dataset is divided into input features and a target variable. The target variable is the selling price, while the remaining selected columns are used to predict it. Categorical variables are encoded before model training.

```python

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Select features and target
features = [
    "name",
    "year",
    "km_driven",
    "fuel",
    "seller_type",
    "transmission",
    "owner"
]

X = df[features].copy()
y = df["selling_price"].copy()

numeric_features = ["year", "km_driven"]

categorical_features = [
    "name",
    "fuel",
    "seller_type",
    "transmission",
    "owner"
]

# Preprocessing pipelines
numeric_transformer = SimpleImputer(strategy="median")

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_transformer, numeric_features),
    ("categorical", categorical_transformer, categorical_features)
])

# Split dataset into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print("Data preparation completed!")
```

Operation: Hypothesis Testing Using One-Way ANOVA
Aim: To determine whether the average selling price differs significantly among different fuel types of used cars.

Null Hypothesis (H₀): The mean selling prices of cars with different fuel types are equal.

Alternative Hypothesis (H₁): At least one fuel type has a different mean selling price.

```python
import pandas as pd
from scipy.stats import f_oneway

# Group car prices by fuel type
groups = [
    group["selling_price"].dropna()
    for _, group in df.groupby("fuel")
    if len(group["selling_price"].dropna()) > 1
]

# Perform ANOVA test
f_stat, p_value = f_oneway(*groups)

print("F-statistic:", f_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0: Significant difference exists.")
else:
    print("Fail to reject H0: No significant difference found.")
```

```python
# ============================================================
# AUTOVAULT - COMPLETE CAR PRICE PREDICTION WEB APP
# Original dark/lime UI + advanced hypothesis testing + accuracy
# ============================================================

import os
import glob
import warnings
warnings.filterwarnings("ignore")

import gradio as gr
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from scipy.stats import f_oneway, kruskal, levene, spearmanr
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import (
    RandomForestRegressor,
    ExtraTreesRegressor,
    GradientBoostingRegressor
)
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# ============================================================
# 1. LOAD DATASET
# ============================================================

possible_files = [
    "CAR DETAILS FROM CAR DEKHO.csv",
    "CAR DETAILS FROM CAR DEKHO(3).csv",
    "CAR DETAILS FROM CAR DEKHO(4).csv",
    "car data.csv",
    "Car details v3.csv",
    "cardekho.csv",
    "car_data.csv",
    "cars.csv"
]

csv_path = next(
    (f for f in possible_files if os.path.isfile(f)),
    None
)

if csv_path is None:
    csvs = glob.glob("*.csv")
    if len(csvs) == 1:
        csv_path = csvs[0]

if csv_path is None:
    raise FileNotFoundError(
        "CarDekho CSV nahi mili. Dataset ko Python file ke same folder "
        "mein rakho, ya notebook ke current folder mein upload karo."
    )

df = pd.read_csv(csv_path)
df.columns = [
    str(c).strip().lower().replace(" ", "_")
    for c in df.columns
]

aliases = {
    "sellingprice": "selling_price",
    "price": "selling_price",
    "km": "km_driven",
    "kms_driven": "km_driven",
    "car_name": "name",
    "model": "name",
    "manufacturing_year": "year",
    "seller": "seller_type",
    "gearbox": "transmission",
    "ownership": "owner"
}

df.rename(
    columns={c: aliases.get(c, c) for c in df.columns},
    inplace=True
)

required = [
    "name", "year", "selling_price", "km_driven",
    "fuel", "seller_type", "transmission", "owner"
]

missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(
        f"Dataset columns missing: {missing}\n"
        f"Available columns: {list(df.columns)}"
    )

for c in ["year", "selling_price", "km_driven"]:
    df[c] = pd.to_numeric(df[c], errors="coerce")

df.dropna(
    subset=["year", "selling_price", "km_driven"],
    inplace=True
)

df = df[df["selling_price"] >= 0].copy()

for c in ["name", "fuel", "seller_type", "transmission", "owner"]:
    df[c] = df[c].fillna("Unknown").astype(str)

if len(df) < 10:
    raise ValueError("Model training ke liye dataset mein enough rows nahi hain.")

features = [
    "name", "year", "km_driven", "fuel",
    "seller_type", "transmission", "owner"
]

X = df[features]
y = df["selling_price"]

# ============================================================
# 2. HELPER FUNCTIONS
# ============================================================

def make_encoder():
    try:
        return OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=True
        )
    except TypeError:
        return OneHotEncoder(
            handle_unknown="ignore",
            sparse=True
        )


def make_pipeline(estimator, data):
    numeric_cols = data.select_dtypes(include=np.number).columns.tolist()
    categorical_cols = [
        c for c in data.columns if c not in numeric_cols
    ]

    transformers = []

    if numeric_cols:
        transformers.append((
            "numeric",
            SimpleImputer(strategy="median"),
            numeric_cols
        ))

    if categorical_cols:
        transformers.append((
            "categorical",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                ("encoder", make_encoder())
            ]),
            categorical_cols
        ))

    prep = ColumnTransformer(transformers=transformers)

    return Pipeline([
        ("preprocessor", prep),
        ("regressor", estimator)
    ])


def money(value):
    value = float(value)

    if not np.isfinite(value):
        return "N/A"

    if abs(value) >= 100000:
        return f"Rs. {value / 100000:.2f} Lakh"

    return f"Rs. {value:,.0f}"


# ============================================================
# 3. TRAIN THE MAIN PREDICTION MODEL
# ============================================================

model = make_pipeline(
    RandomForestRegressor(
        n_estimators=150,
        random_state=42,
        min_samples_leaf=2,
        n_jobs=-1
    ),
    X
)

model.fit(X, y)

# ============================================================
# 4. DROPDOWN OPTIONS
# ============================================================

car_names = sorted(df["name"].unique().tolist())
fuel_options = sorted(df["fuel"].unique().tolist())
seller_options = sorted(df["seller_type"].unique().tolist())
transmission_options = sorted(df["transmission"].unique().tolist())
owner_options = sorted(df["owner"].unique().tolist())

min_year = int(df["year"].min())
max_year = max(2026, int(df["year"].max()))

total_cars = len(df)
total_brands = df["name"].nunique()
avg_price = df["selling_price"].mean()

# ============================================================
# 5. PRICE PREDICTION
# ============================================================

def predict_price(name, year, km, fuel, seller, transmission, owner):

    if not name:
        return "### Please select a car model."

    if km is None or not np.isfinite(float(km)) or float(km) < 0:
        return "### Please enter valid kilometres driven."

    input_data = pd.DataFrame([{
        "name": name,
        "year": int(year),
        "km_driven": int(km),
        "fuel": fuel,
        "seller_type": seller,
        "transmission": transmission,
        "owner": owner
    }])[features]

    prediction = max(0, float(model.predict(input_data)[0]))

    return f"""
# {money(prediction)}

**Estimated Resale Price**

| Vehicle Detail | Selected Value |
|---|---|
| Car Model | {name} |
| Manufacturing Year | {int(year)} |
| Kilometres Driven | {int(km):,} km |
| Fuel | {fuel} |
| Seller Type | {seller} |
| Transmission | {transmission} |
| Ownership | {owner} |

*This is a machine-learning estimate, not a guaranteed market price.*
"""

# ============================================================
# 6. ADVANCED HYPOTHESIS TESTING
# ============================================================

def hypothesis_report():

    groups = []
    group_names = []

    for fuel_name, group in df.groupby("fuel"):
        values = group["selling_price"].dropna().to_numpy()
        values = values[np.isfinite(values)]

        if len(values) >= 2:
            groups.append(values)
            group_names.append(str(fuel_name))

    sections = [
        "## Advanced Hypothesis Testing",
        "",
        "**Significance level:** 0.05",
        "",
        "### Test 1: One-Way ANOVA",
        "",
        "**H0:** All fuel groups have the same mean selling price.",
        "",
        "**H1:** At least one fuel group has a different mean selling price.",
        ""
    ]

    if len(groups) >= 2:
        f_stat, anova_p = f_oneway(*groups)
        kw_stat, kw_p = kruskal(*groups)
        lev_stat, lev_p = levene(*groups, center="median")

        all_prices = np.concatenate(groups)
        grand_mean = np.mean(all_prices)

        ss_between = sum(
            len(g) * (np.mean(g) - grand_mean) ** 2
            for g in groups
        )

        ss_total = sum(
            np.sum((g - grand_mean) ** 2)
            for g in groups
        )

        eta_squared = (
            ss_between / ss_total if ss_total > 0 else 0
        )

        sections.extend([
            "| Statistical Measure | Result |",
            "|---|---:|",
            f"| Fuel groups | {len(groups)} |",
            f"| ANOVA F-statistic | {f_stat:.4f} |",
            f"| ANOVA p-value | {anova_p:.6g} |",
            f"| Kruskal-Wallis H-statistic | {kw_stat:.4f} |",
            f"| Kruskal-Wallis p-value | {kw_p:.6g} |",
            f"| Levene variance-test p-value | {lev_p:.6g} |",
            f"| Eta-squared effect size | {eta_squared:.4f} |",
            "",
            "**ANOVA conclusion:** " + (
                "Reject H0: at least one group mean differs significantly."
                if anova_p < 0.05 else
                "Fail to reject H0: insufficient evidence of different group means."
            ),
            "",
            "**Kruskal-Wallis conclusion:** " + (
                "Reject H0: at least one group's rank distribution differs."
                if kw_p < 0.05 else
                "Fail to reject H0: insufficient evidence of distribution differences."
            ),
            "",
            "Eta-squared estimates the proportion of observed price variation "
            "associated with fuel-group membership in this sample.",
            ""
        ])
    else:
        sections.append("Not enough fuel groups for ANOVA and Kruskal-Wallis.")

    sections.extend([
        "### Test 2: Spearman Correlation",
        "",
        "**H0:** There is no monotonic relationship between the feature and price.",
        ""
    ])

    for col, label in [
        ("year", "Manufacturing year"),
        ("km_driven", "Kilometres driven")
    ]:
        temp = df[[col, "selling_price"]].apply(
            pd.to_numeric, errors="coerce"
        ).dropna()

        if len(temp) >= 3 and temp[col].nunique() > 1:
            rho, p_value = spearmanr(
                temp[col], temp["selling_price"]
            )

            sections.append(
                f"- **{label}:** Spearman rho = {rho:.4f}, "
                f"p-value = {p_value:.6g}. "
                + (
                    "Statistically significant at 0.05."
                    if p_value < 0.05 else
                    "Not statistically significant at 0.05."
                )
            )

    sections.extend([
        "",
        "### Interpretation",
        "- ANOVA compares group means.",
        "- Kruskal-Wallis provides a rank-based comparison.",
        "- Levene's test checks differences in group variances.",
        "- Correlation does not prove causation.",
        "- Statistical significance does not automatically imply a large practical effect."
    ])

    return "\n".join(sections)


advanced_hypothesis = hypothesis_report()

# ============================================================
# 7. FAIR MODEL COMPARISON ON A HELD-OUT TEST SET
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

candidates = {
    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        min_samples_leaf=2,
        n_jobs=-1
    ),
    "Extra Trees": ExtraTreesRegressor(
        n_estimators=100,
        random_state=42,
        min_samples_leaf=2,
        n_jobs=-1
    ),
    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=100,
        random_state=42,
        learning_rate=0.05
    )
}

comparison_rows = []

for model_name, estimator in candidates.items():

    candidate_model = make_pipeline(estimator, X_train)
    candidate_model.fit(X_train, y_train)

    predictions = candidate_model.predict(X_test)

    actual = np.asarray(y_test, dtype=float)
    predicted = np.asarray(predictions, dtype=float)

    mae = mean_absolute_error(actual, predicted)
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    r2 = r2_score(actual, predicted)

    positive = actual > 0

    if positive.any():
        mape = np.mean(
            np.abs(
                (actual[positive] - predicted[positive])
                / actual[positive]
            )
        ) * 100

        accuracy_10 = np.mean(
            np.abs(
                (actual[positive] - predicted[positive])
                / actual[positive]
            ) <= 0.10
        ) * 100
    else:
        mape = np.nan
        accuracy_10 = np.nan

    comparison_rows.append({
        "Model": model_name,
        "MAE (Rs.)": round(float(mae), 2),
        "RMSE (Rs.)": round(float(rmse), 2),
        "R2 Score": round(float(r2), 4),
        "MAPE (%)": round(float(mape), 2),
        "Within +/-10% (%)": round(float(accuracy_10), 2)
    })

comparison_df = pd.DataFrame(comparison_rows).sort_values(
    "RMSE (Rs.)"
).reset_index(drop=True)

best = comparison_df.iloc[0]

best_model_name = best["Model"]
best_r2 = best["R2 Score"]
best_rmse = best["RMSE (Rs.)"]
best_mae = best["MAE (Rs.)"]
best_mape = best["MAPE (%)"]
best_accuracy = best["Within +/-10% (%)"]

# ============================================================
# 8. GRAPHS
# ============================================================

def style_plot(fig, ax):
    fig.patch.set_facecolor("#191c17")
    ax.set_facecolor("#191c17")
    ax.tick_params(colors="#d9ddd1")
    ax.xaxis.label.set_color("#d9ddd1")
    ax.yaxis.label.set_color("#d9ddd1")
    ax.title.set_color("#d6ff3f")

    for spine in ax.spines.values():
        spine.set_color("#50564a")

    ax.grid(axis="y", alpha=0.2)


def plot_fuel_prices():
    summary = df.groupby("fuel")["selling_price"].mean().sort_values(
        ascending=False
    )

    fig, ax = plt.subplots(figsize=(8, 4))
    style_plot(fig, ax)

    ax.bar(summary.index.astype(str), summary.values, color="#d6ff3f")
    ax.set_title("Average Selling Price by Fuel Type")
    ax.set_xlabel("Fuel Type")
    ax.set_ylabel("Average Selling Price (Rs.)")
    ax.tick_params(axis="x", rotation=15)

    fig.tight_layout()
    return fig


def plot_year_prices():
    summary = df.groupby("year")["selling_price"].median().sort_index()

    fig, ax = plt.subplots(figsize=(8, 4))
    style_plot(fig, ax)

    ax.plot(summary.index, summary.values, marker="o", color="#d6ff3f")
    ax.set_title("Median Selling Price by Manufacturing Year")
    ax.set_xlabel("Manufacturing Year")
    ax.set_ylabel("Median Selling Price (Rs.)")

    fig.tight_layout()
    return fig


def plot_model_rmse():
    fig, ax = plt.subplots(figsize=(8, 4))
    style_plot(fig, ax)

    ordered = comparison_df.sort_values("RMSE (Rs.)")
    bars = ax.bar(
        ordered["Model"], ordered["RMSE (Rs.)"], color="#d6ff3f"
    )

    if len(bars):
        bars[0].set_color("#a8d12c")

    ax.set_title("Model Comparison - Lower RMSE Is Better")
    ax.set_ylabel("RMSE (Rs.)")
    ax.tick_params(axis="x", rotation=12)

    fig.tight_layout()
    return fig


# ============================================================
# 9. ORIGINAL-STYLE AUTOVAULT UI
# ============================================================

custom_css = """
@import url('https://fonts.googleapis.com/css2?family=DM+Sans:wght@400;500;600;700&family=Space+Grotesk:wght@400;500;600;700&display=swap');

:root {
    --lime: #d6ff3f;
    --dark: #10120f;
    --panel: #191c17;
}

.gradio-container {
    background: var(--dark) !important;
    color: #f4f5ef !important;
    font-family: 'DM Sans', sans-serif !important;
    max-width: 1440px !important;
    margin: auto !important;
}

.gradio-container h1,
.gradio-container h2,
.gradio-container h3 {
    font-family: 'Space Grotesk', sans-serif !important;
    color: #f5f6f0 !important;
}

#main-header {
    border: 1px solid #34382e;
    border-radius: 24px;
    overflow: hidden;
    background:
        linear-gradient(90deg, rgba(13,16,12,.98), rgba(13,16,12,.20)),
        url('https://images.unsplash.com/photo-1492144534655-ae79c964c9d7?auto=format&fit=crop&w=1600&q=85')
        center / cover;
    padding: 48px 38px;
    min-height: 280px;
}

.eyebrow {
    color: var(--lime);
    font-size: 11px;
    font-weight: 700;
    letter-spacing: 3px;
}

.hero-title {
    font-size: clamp(36px, 5vw, 66px);
    font-weight: 700;
    letter-spacing: -2px;
    line-height: 1.05;
    margin: 18px 0;
}

.hero-title span {
    color: var(--lime);
}

.hero-subtitle {
    color: #c1c5bb;
    font-size: 14px;
    line-height: 1.8;
}

.hero-tag {
    display: inline-block;
    margin-top: 18px;
    border: 1px solid #4a523b;
    border-radius: 30px;
    padding: 9px 15px;
    color: var(--lime);
    font-size: 12px;
}

.stat-card {
    background: linear-gradient(145deg, #20241c, #151712);
    border: 1px solid #34382e;
    border-radius: 17px;
    padding: 19px;
    min-height: 88px;
}

.stat-label {
    font-size: 11px;
    color: #9da396;
    text-transform: uppercase;
}

.stat-value {
    color: var(--lime);
    font-size: 24px;
    font-weight: 700;
    margin-top: 8px;
    overflow-wrap: anywhere;
}

.section-heading {
    border-bottom: 1px solid #35392f;
    padding: 8px 0 14px;
    margin: 22px 0 18px;
}

.section-heading p {
    color: #a4a99d;
    font-size: 13px;
}

.panel {
    background: #191c17;
    border: 1px solid #34382e;
    border-radius: 20px;
    padding: 23px;
}

.gradio-container label {
    color: #d9ddd1 !important;
}

.gradio-container input,
.gradio-container textarea {
    background: #11130f !important;
    border-color: #383d32 !important;
    border-radius: 10px !important;
}

#predict-btn {
    background: var(--lime) !important;
    color: #14170e !important;
    border: none !important;
    border-radius: 12px !important;
    min-height: 52px;
    font-weight: 800 !important;
}

#result-panel {
    background: linear-gradient(145deg, #22271a, #151811);
    border: 1px solid #505a37;
    border-radius: 20px;
    padding: 23px;
    min-height: 220px;
}

#result-panel h1,
#result-panel h2 {
    color: var(--lime) !important;
    overflow-wrap: anywhere;
}

.footer-note {
    text-align: center;
    color: #82877b;
    font-size: 11px;
    padding: 22px 0 8px;
}

"""

with gr.Blocks(
    title="AUTOVAULT | Car Price Intelligence",
    css=custom_css,
    theme=gr.themes.Base()
) as demo:

    gr.HTML("""
    <div id="main-header">
        <div class="eyebrow">AUTOVAULT / INTELLIGENCE LAB</div>
        <div class="hero-title">
            Find the car.<br>Know the <span>value.</span>
        </div>
        <div class="hero-subtitle">
            A smarter way to estimate used-car prices.
            Powered by machine learning and real dataset patterns.
        </div>
        <div class="hero-tag">AI-POWERED PRICE ESTIMATION</div>
    </div>
    """)

    with gr.Row():
        gr.HTML(f"""
        <div class="stat-card">
            <div class="stat-label">Dataset records</div>
            <div class="stat-value">{total_cars:,}</div>
        </div>
        """)
        gr.HTML(f"""
        <div class="stat-card">
            <div class="stat-label">Car variants</div>
            <div class="stat-value">{total_brands:,}</div>
        </div>
        """)
        gr.HTML(f"""
        <div class="stat-card">
            <div class="stat-label">Average listed price</div>
            <div class="stat-value">Rs. {avg_price / 100000:.2f} Lakh</div>
        </div>
        """)

    gr.HTML("""
    <div class="section-heading">
        <div class="eyebrow">01 / CONFIGURE YOUR VEHICLE</div>
        <h2 style="margin:8px 0 0;font-size:27px;">Build your car profile</h2>
        <p>Enter vehicle specifications to estimate its resale value.</p>
    </div>
    """)

    with gr.Row(equal_height=False):

        with gr.Column(scale=6, elem_classes=["panel"]):
            gr.Markdown("### Vehicle identity")

            name = gr.Dropdown(
                choices=car_names,
                value=car_names[0],
                label="CAR MODEL",
                filterable=True
            )

            with gr.Row():
                year = gr.Slider(
                    minimum=min_year,
                    maximum=max_year,
                    value=min(max_year, max(min_year, 2018)),
                    step=1,
                    label="MANUFACTURING YEAR"
                )

                km = gr.Number(
                    value=50000,
                    precision=0,
                    label="KILOMETRES DRIVEN"
                )

            gr.Markdown("### Vehicle specifications")

            with gr.Row():
                fuel = gr.Dropdown(
                    choices=fuel_options,
                    value=fuel_options[0],
                    label="FUEL TYPE"
                )

                transmission = gr.Dropdown(
                    choices=transmission_options,
                    value=transmission_options[0],
                    label="TRANSMISSION"
                )

            with gr.Row():
                seller = gr.Dropdown(
                    choices=seller_options,
                    value=seller_options[0],
                    label="SELLER TYPE"
                )

                owner = gr.Dropdown(
                    choices=owner_options,
                    value=owner_options[0],
                    label="OWNERSHIP HISTORY"
                )

            predict_button = gr.Button(
                "ESTIMATE CAR VALUE",
                variant="primary",
                elem_id="predict-btn"
            )

        with gr.Column(scale=5, elem_id="result-panel"):
            gr.HTML("""
            <div class="eyebrow">02 / YOUR VALUATION</div>
            <h3 style="font-size:25px;margin:10px 0;">Price intelligence</h3>
            <p style="color:#a4a99d;font-size:13px;line-height:1.8;">
                Your estimated resale value and vehicle summary appear here.
            </p>
            """)

            output = gr.Markdown(
                "### Ready when you are\nSelect the vehicle details and click Estimate Car Value."
            )

    predict_button.click(
        fn=predict_price,
        inputs=[name, year, km, fuel, seller, transmission, owner],
        outputs=output
    )

    # ANALYTICS FEATURES ARE PART OF THE UI
    gr.HTML("""
    <div class="section-heading">
        <div class="eyebrow">03 / MODEL INTELLIGENCE</div>
        <h2 style="margin:8px 0 0;font-size:27px;">Analytics Workspace</h2>
        <p>Hypothesis tests, model evaluation and used-car market insights.</p>
    </div>
    """)

    with gr.Row():
        gr.HTML(f"""
        <div class="stat-card">
            <div class="stat-label">Best evaluated model</div>
            <div class="stat-value">{best_model_name}</div>
        </div>
        """)
        gr.HTML(f"""
        <div class="stat-card">
            <div class="stat-label">R2 score</div>
            <div class="stat-value">{best_r2:.3f}</div>
        </div>
        """)
        gr.HTML(f"""
        <div class="stat-card">
            <div class="stat-label">Within +/-10% price error</div>
            <div class="stat-value">{best_accuracy:.1f}%</div>
        </div>
        """)
        gr.HTML(f"""
        <div class="stat-card">
            <div class="stat-label">RMSE</div>
            <div class="stat-value">{money(best_rmse)}</div>
        </div>
        """)

    with gr.Tabs():

        with gr.Tab("Hypothesis Lab"):
            gr.Markdown(advanced_hypothesis)

            with gr.Row():
                gr.Plot(
                    value=plot_fuel_prices,
                    label="Average price by fuel type"
                )
                gr.Plot(
                    value=plot_year_prices,
                    label="Price trend by manufacturing year"
                )

        with gr.Tab("Model Accuracy"):
            gr.Markdown("""
            ## Model Performance Comparison

            **R2** is a regression score, not classification accuracy.

            **Within +/-10%** reports the percentage of positive-price test
            examples whose predicted price is within 10% of the actual price.
            """)

            gr.Dataframe(
                value=comparison_df,
                interactive=False,
                label="Test-set model metrics"
            )

            gr.Plot(
                value=plot_model_rmse,
                label="RMSE comparison"
            )

            gr.Markdown(f"""
            ### Best model by lowest RMSE: {best_model_name}

            - **MAE:** {money(best_mae)}
            - **RMSE:** {money(best_rmse)}
            - **R2 score:** {best_r2:.4f}
            - **MAPE:** {best_mape:.2f}%
            - **Within +/-10%:** {best_accuracy:.2f}%

            The model comparison uses a held-out test set. The main prediction
            panel continues using the trained Random Forest model.
            """)

        with gr.Tab("Market Insights"):
            gr.Markdown(f"""
            ## Dataset Overview

            - **Records analysed:** {total_cars:,}
            - **Distinct car variants:** {total_brands:,}
            - **Average selling price:** {money(avg_price)}
            - **Fuel categories:** {len(fuel_options)}
            - **Year range in dataset:** {min_year} to {int(df["year"].max())}
            """)

            with gr.Row():
                gr.Plot(value=plot_fuel_prices)
                gr.Plot(value=plot_year_prices)

    gr.HTML("""
    <div class="footer-note">
        AUTOVAULT 2026 | PYTHON + MACHINE LEARNING | EDUCATIONAL PROJECT
    </div>
    """)

# ============================================================
# 10. LAUNCH
# ============================================================

demo.launch(share=True)
```
